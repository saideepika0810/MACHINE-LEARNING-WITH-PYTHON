```python
import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder

data = pd.read_csv("experiment_1.csv")

# Encode data
enc = {}
for col in data.columns:
    enc[col] = LabelEncoder()
    data[col] = enc[col].fit_transform(data[col])

X = data.drop("EnjoySport", axis=1)
y = data["EnjoySport"]

# Train model
model = GaussianNB()
model.fit(X, y)

# Predict
pred = model.predict(X)

for i in range(len(y)):
    actual = enc["EnjoySport"].inverse_transform([y[i]])[0]
    predicted = enc["EnjoySport"].inverse_transform([pred[i]])[0]
    print("Actual:", actual, "Predicted:", predicted)

print("\nAccuracy:", (pred == y).mean() * 100, "%")
```
