```python
import pandas as pd
from sklearn.neural_network import MLPClassifier

# Read training data
data = pd.read_csv("experiment_5.csv")

# Separate input and output
X = data[["Input1", "Input2"]]
y = data["Output"]

# Create Artificial Neural Network
model = MLPClassifier(
    hidden_layer_sizes=(4,),
    activation="logistic",
    solver="lbfgs",
    max_iter=1000,
    random_state=1
)

# Train the network using Back Propagation
model.fit(X, y)

# Predict output
predictions = model.predict(X)

# Display results
print("Input1 Input2 Actual Predicted")

for i in range(len(X)):
    print(
        X.iloc[i, 0],
        "      ",
        X.iloc[i, 1],
        "      ",
        y.iloc[i],
        "       ",
        predictions[i]
    )
```
