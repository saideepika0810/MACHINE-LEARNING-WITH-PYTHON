```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score, adjusted_rand_score

data = pd.read_csv("experiment_9.csv")

X = data[["Feature1", "Feature2"]]

# K-Means clustering
km = KMeans(n_clusters=3, random_state=42, n_init=10)
klabels = km.fit_predict(X)

# EM clustering
em = GaussianMixture(n_components=3, random_state=42)
em.fit(X)
elabels = em.predict(X)

# Compare results
print("K-Means Labels:", klabels.tolist())
print("EM Labels:", elabels.tolist())

print("K-Means Silhouette:",
      round(silhouette_score(X, klabels), 3))
print("EM Silhouette:",
      round(silhouette_score(X, elabels), 3))

print("K-Means ARI:",
      round(adjusted_rand_score(data["KnownGroup"], klabels), 3))
print("EM ARI:",
      round(adjusted_rand_score(data["KnownGroup"], elabels), 3))
```
