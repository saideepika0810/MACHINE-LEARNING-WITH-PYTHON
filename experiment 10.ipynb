```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv("experiment_10.csv")
x = data["X"].values
y = data["Y"].values

def lwr(x, y, point, tau=1.5):
    w = np.exp(-((x - point)**2) / (2 * tau**2))
    X = np.c_[np.ones(len(x)), x]
    W = np.diag(w)
    b = np.linalg.pinv(X.T @ W @ X) @ X.T @ W @ y
    return np.array([1, point]) @ b

points = np.linspace(min(x), max(x), 200)
predictions = [lwr(x, y, p) for p in points]

plt.scatter(x, y, label="Actual Data")
plt.plot(points, predictions, label="LWR Curve")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Locally Weighted Regression")
plt.legend()
plt.show()
```
