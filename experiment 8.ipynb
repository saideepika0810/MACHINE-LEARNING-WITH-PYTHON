```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, precision_score, recall_score

data = pd.read_csv("experiment_8.csv")

X = CountVectorizer().fit_transform(data["Document"])
y = data["Class"]

model = MultinomialNB()
model.fit(X, y)
pred = model.predict(X)

for d, a, p in zip(data["Document"], y, pred):
    print(d, "-> Actual:", a, "Predicted:", p)

print("Accuracy:", accuracy_score(y, pred) * 100, "%")
print("Precision:", precision_score(y, pred, pos_label="Spam") * 100, "%")
print("Recall:", recall_score(y, pred, pos_label="Spam") * 100, "%")
```
