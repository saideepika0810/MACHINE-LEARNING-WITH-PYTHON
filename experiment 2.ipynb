```python
import pandas as pd

# Read training data from CSV file
data = pd.read_csv("experiment_1.csv")

attributes = list(data.columns[:-1])

# Initialize Specific Boundary
S = ['Ø'] * len(attributes)

# Initialize General Boundary
G = ['?'] * len(attributes)

# Candidate-Elimination Algorithm
for _, row in data.iterrows():

    example = list(row.iloc[:-1])
    target = row.iloc[-1]

    if target == 'Yes':

        # Generalize S
        for i in range(len(attributes)):
            if S[i] == 'Ø':
                S[i] = example[i]
            elif S[i] != example[i]:
                S[i] = '?'

    else:

        # Specialize G to exclude negative example
        for i in range(len(attributes)):
            if S[i] != '?' and S[i] != example[i]:
                G[i] = S[i]

print("Specific Boundary (S):",S)


print("\nGeneral Boundary (G):",G)
```
